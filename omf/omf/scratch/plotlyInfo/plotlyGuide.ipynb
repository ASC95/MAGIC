```python
import plotly

# Necessary for using in jupyter notebooks to avoid 25 count limit with your plotly account.
plotly.offline.init_notebook_mode(connected=True)

# The offline plot() function will create an html file with a plot. Use iplot() wihtin notebooks.
from plotly.offline import plot, iplot
import plotly.graph_objs as go

# Create a trace of points
trace = go.Scatter(
    x = [0,1,2,3,4],
    y = [5,2,7,8,3]
)

data = [trace]

# ipython notebook plotting function is used below.
# To generate an html file, use the plotly.offline.plot() function.
# For omf models, use the plotly json serializer. json.dumps(figure, cls=plotly.utils.PlotlyJSONEncoder)
plotly.offline.iplot(data)
```

```python
# Plot multiple data sets on the same plot

# Create random data with numpy
import numpy as np

N = 100
random_x = np.linspace(0, 1, N)
random_y0 = np.random.randn(N)
random_y1 = np.random.randn(N)
random_y2 = np.random.randn(N)

# Create three different traces, each 100 random y values along the x axis.
# The mode determines how the data is displayed.
# The name will show up on the side with the associated trace. Click this to show/hide a trace.

trace0 = go.Scatter(
    x = random_x,
    y = random_y0,
    mode = 'markers',
    name = 'markers'
)
trace1 = go.Scatter(
    x = random_x,
    y = random_y1,
    mode = 'lines+markers',
    name = 'lines+markers'
)
trace2 = go.Scatter(
    x = random_x,
    y = random_y2,
    mode = 'lines',
    name = 'lines'
)

#Plot multiple traces on a single graph. The data on your plotly graph is an array of traces.
data = [trace0, trace1, trace2]

#Offline plotting function
iplot(data)
```

```python
# Create a layout object.
# Set the height and width. The height and width sets the container not only for the graph, but also the title and legends. 
# Keep this in mind when embedding plotly plots in your templates.
# Label your plots and axes.

layout = go.Layout(
    width = 800,
    height = 500,
    title = 'Plot Title',
    xaxis = dict(
        title = 'X Axis'
        ),
    yaxis = dict(
        title = 'Y Axis',
        titlefont = dict(
            family='Courier New, monospace',
            size=30,
            color='#0000FF')
        )
)
# The figure object takes both the traces in the data array and the layout object.
fig = go.Figure(data=data, layout=layout)

#Offline plotting function. Using the figure object, combination of data and layout.
iplot(fig)
```

```python
# Use plotly to create a table directly from your dataset and minimize additional work on the front end.
# The table object is an array of column headers and arrays of values for each column.
trace = go.Table(
    header=dict(values=['A Scores', 'B Scores'],
                line = dict(color='#7D7F80'),
                fill = dict(color='#a1c3d1'),
                align = ['left'] * 5),
    cells=dict(values=[[100, 90, 80, 90],
                       [95, 85, 75, 95]],
               line = dict(color='#7D7F80'),
               fill = dict(color='#EDFAFF'),
               align = ['left'] * 5))

layout = dict(width=500, height=300)

data = [trace]

fig = dict(data=data, layout=layout)

iplot(fig)
```

```python
# Create scatter plots
# Tis example dispalys a scatter plot with 1000 random points
N = 1000
random_x = np.random.randn(N)
random_y = np.random.randn(N)

# Create a trace
trace = go.Scatter(
    x = random_x,
    y = random_y,
    mode = 'markers',
)

data = [trace]

iplot(data)

# For larger values, use the Scattergl type.
N = 100000
trace = go.Scattergl(
    x = np.random.randn(N),
    y = np.random.randn(N),
    mode = 'markers',
)
data = [trace]

iplot(data)
```

```python
# In bar charts, the x value is the group/category, and the y value is the count in that group/category.

trace1 = go.Bar(
            x=['giraffes', 'orangutans', 'monkeys'],
            y=[20, 14, 23],
            name='SF Zoo'
    )
data = [trace1]

# Plot basic bar chart here

# Add another trace to the bar chart
trace2 = go.Bar(
    x=['giraffes', 'orangutans', 'monkeys'],
    y=[12, 18, 29],
    name='LA Zoo'
)

data = [trace1, trace2]

#plot(data)

# Set the barmaode value to stack the data on top of one another
layout = go.Layout(
    barmode='stack'
)

fig = go.Figure(data=data, layout=layout)

iplot(fig)
```

```python
# Histograms automatically bin values for you.

x = np.random.randn(500)
trace = go.Histogram(x=x)

data = [trace]

iplot(data)

import numpy as np

# Create an overlay hsitogram with two traces.
x0 = np.random.randn(500)
x1 = np.random.randn(500)+1

trace1 = go.Histogram(
    x=x0,
    opacity=0.7
)
trace2 = go.Histogram(
    x=x1,
    opacity=0.7
)

data = [trace1, trace2]

# Set the mode to overlay. Alternatively, set the barmode value to stack 
layout = go.Layout(
	barmode='overlay')

fig = go.Figure(data=data, layout=layout)
iplot(fig)
```

```python
import pandas as pd

# This example shows using a pandas dataframe to plot time series data

df = pd.read_csv("three_month_data.csv")

# See the last 5 entries in the pandas dataframe. There are two columns, date and temperature.
df.tail()

# The x an y values are populated with columns from the pandas dataframe.
trace = go.Scatter(
    y=df['temperature'], 
    x=df['date'], 
    name=('Charlottesville'), 
    marker=dict(color='red'))

data = [trace]

# This layout changes the legend placement and display. 
# Note, if there is only one trace, the showlegend value must be set to True. If there are multiple traces, the legends are displayed by default.
layout = go.Layout(
    title='Temperature (F)',
    showlegend=True, 
    legend=dict(
        x=0, 
        y=1.2,
        bgcolor='#E2E2E2',
        bordercolor='#FFFFFF',
        borderwidth=2), 
    height=400, width=980)

fig = dict(data=data, layout=layout)

iplot(fig)
```

```python
#This example shows how to use plotly to create heatmaps.
trace = go.Heatmap(z=[[1, 20, 30, 50, 1], [20, 1, 60, 80, 30], [30, 60, 1, -10, 20]],
                   x=['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday'],
                   y=['Morning', 'Afternoon', 'Evening'])
data=[trace]
iplot(data)
```

```python
#Create a small multiples chart using subplots.
from plotly import tools

#Create your figure as shown below. You will add subplots to this figure
#Define the layout fo the subplots with the rows and columns values 
fig = tools.make_subplots(rows=2, cols=2)

df = pd.read_csv("household_watts.csv")
df.set_index('date', inplace=True)

df.tail()

fig.append_trace(go.Scatter(y=df['Massachusetts'], x=df.index, name=('Massachusetts')),1,1)
fig.append_trace(go.Scatter(y=df['Texas'], x=df.index, name=('Texas')),1,2)
fig.append_trace(go.Scatter(y=df['Kentucky'], x=df.index, name=('Kentucky')),2,1)
fig.append_trace(go.Scatter(y=df['Nevada'], x=df.index, name=('Nevada')),2,2)

#Define a title for the overall plot
fig['layout'].update(title='Average Consumption by Household')

#Label the y axis for each subplot
fig['layout']['yaxis1'].update(title='kWh')
fig['layout']['yaxis2'].update(title='kWh')
fig['layout']['yaxis3'].update(title='kWh')
fig['layout']['yaxis4'].update(title='kWh')

# Standardize the y axis 
fig['layout']['yaxis'].update(range=[700,1000])
fig['layout']['yaxis2'].update(range=[700,1000])
fig['layout']['yaxis3'].update(range=[700,1000])
fig['layout']['yaxis4'].update(range=[700,1000])

iplot(fig)
```
